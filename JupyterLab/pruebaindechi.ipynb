- La prueba de Chi-Cuadrado
- X^2 
- Es una de ls herramientas de estadística indeferencial. Se utiliza para analizar variables categóricas, aquellas que clasifican a los individuos con grupos o categorías.
- Variables categóricas:
    - Sexo, nivel educativo, preferencia de marca o grupo de edad.
- Objetivo: Determinar si existe una relación estadísticamente significativa entre dos o más variables.


## Corazón de la prueba:
- Frecuencias Observadas (O)
    - Los datos reales recogidos en la muestra, tal como aparecen en la tabla de contingencia.
- Frecuencias Esperadas (E)
    - Los valores que predice la teoría si no existiera ninguna relación entre las variables.
- La desviación importa
    - Si O y E difieren significativamente, hay evidencia de una relación o patrón no aleatorio.


## Hipótesis clave: El fundamento de la prueba
- Nula: No existe relación entre las variables categóricas analizadas. Cualquier diferencia observada se debe únicamente al azar.
    - No existe relación
- Alternativa: Sí eiste una relación entre las variables categóricas. Las diferencias observadas son estadísticamente significativas y no un producto al azar.
    - Si existe relación

## Tipos de pruebas de chi - cuadrado

- Cada varibale responde a una pregunta estadística distinta. Conocer cual aplicar es clave.

### Prueba de independencia
 - ¿Son dos variables categoricas independientes entre sí?
    - Analiza la distribución de una variable cambia segpun los niveles de otra, revelando posibles asociaciones.
    - Ejemplo clásico: ¿Influye el sexo en la probabilidad de suscribirse a un servicio de streaming como Netflix?, Se comparan las frecuencias de suscripción entre hombres y mujeres.

```python
import numpy as np
from scipy import stats
# Tabla de frecuencias observadas.
tabla = np.array(
    [[40,20],
    [30,30]])
# Realiza la prueba de chi-cuadrado de independencia stats.chi2_contigency devuelve el estadístico
# chi-cuadrado, p-value, los grados de libertad y las  frecuencias esperadas.abs

chi2, p_value, gl, esperados = stats.chi2_contingency(tabla, correction=False)

# Por alguna razón ponemos correction=false para arreglos de 2x2

print(f"Chi-cuadrado = {chi2:.4f}")
print(f"P-value = {p_value:.4f}")
print(f"Grados de libertad = {gl}")
print(f"Frecuenicas esperadas = {esperados}")

alpha = 0.05

if p_value < alpha: 
    #print("Rechazo H0: No hay una asociación significativa entre las variables.")
    print("Hay evidencia suficiente para determinar que hacer ejercicio depende del género.")
else:
    #print("No rechazo H0: No hay evidencia suficiente de la asociacio significtiva entre las variables")
    print("No hay evidencia suficiente para determinar que hacer ejercicio depende del género.")
```

## Fórmula detrás del poder.
- Entender la mecanica.
### Cuando el valor de chi-cuadrado (X^2)
- Fórmula central:  $$X^2= \sum \frac{(O-E)^2}{E}$$
- Donde O son las frecuencias observadas y E son las frecuencias esperadas para cada categoría.
- Registrar O, Calcular E, Aplicar (O-E)^2/E, Sumar X^2.
### X^2 Alto
- Gran diferencia O vs E -> Evidencia de relación.
### X^2 Bajo
- O=E -> Variables probablemente independientes.

### Frecuencias Observadas vs Esperadas.
- Tabla de contingenci organiza los datos en filas y columnas para cada combinación de categorías, permitiendo calculat las frecuencias esperadas bajo independiencia.

|-|-|-|-|
|Categoría|Observado(O)|Esperado(E)|(O-E)^2/E|
|Grupo A|45|38.5|1.098|
|Grupo B|30|36.5|1.157|
|Grupo C|25|25.0|0.000|
|Total|100|100|X^2=2.255|

- Cuanto mayor sea la suma final, más evidencia existe para la hipótesis nula.
